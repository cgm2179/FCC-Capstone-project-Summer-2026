### Validate + export

```python
import fw_unet2d, fw_infer, fw_export
model = fw_unet2d.load_model('fw_unet2d_indoor.pt')
# g2 gate: tiled prediction vs a FRESH FDTD solve on a held-out Tx
res = fw_infer.validate(model, 'LTE_B71_617', seed=99)
print(res)
# export ONNX + browser contract
fw_export.export_onnx(model, fw_export.WEB / 'fw_unet2d.onnx')
print('exported', fw_export.WEB / 'fw_unet2d.onnx')
```
